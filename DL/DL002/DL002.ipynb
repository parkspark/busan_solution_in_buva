````python
import numpy as np
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
def sigmoid_deriv(z):
    s = sigmoid(z)
    return s * (1 - s)
z_vals = np.linspace(-5, 5, 1000)
max_deriv = sigmoid_deriv(z_vals).max()
print(f"Sigmoid 도함수 최대값: {max_deriv:.4f}")

# Sigmoid 도함수 최대값: 0.2500
````
